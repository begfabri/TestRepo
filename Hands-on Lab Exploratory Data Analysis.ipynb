#### Step 2: Load and Preview the Dataset
Load the dataset from the provided URL. Use df.head() to display the first few rows to get an overview of the structure.


```python
# Load the Stack Overflow survey dataset
data_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv'
df = pd.read_csv(data_url)

# Set pandas option to display all columns
pd.set_option('display.max_columns', None)

# Display the first few rows of the dataset
df.head()
```

#### Step 3: Handling Missing Data


Identify and manage missing values in critical columns such as `Employment`, `JobSat`, and `RemoteWork`. Implement a strategy to fill or drop these values, depending on the significance of the missing data.


```python
## Write your code here
# Work on a copy so the raw dataset remains unchanged
df_eda = df.copy()

# Convert numeric analysis columns where possible
for column in ['JobSat', 'JobSatPoints_1']: # Iterate over the specified columns
    if column in df_eda.columns:
        df_eda[column] = pd.to_numeric(df_eda[column], errors='coerce') # Convert to numeric, coercing errors to NaN

# Impute categorical critical fields with their mode
for column in ['Employment', 'RemoteWork']:
    if column in df_eda.columns:
        mode = df_eda[column].mode(dropna=True) # Get the mode of the column, ignoring NaN values
        if not mode.empty:
            df_eda[column] = df_eda[column].fillna(mode.iloc[0]) # Fill NaN values with the mode of the column

# Impute JobSat with the median because it is an ordinal numeric score.
if 'JobSat' in df_eda.columns:
    job_sat_median = df_eda['JobSat'].median() # Get the median of the column
    df_eda['JobSat'] = df_eda['JobSat'].fillna(job_sat_median) # Fill NaN values with the median of the column  

print(df_eda[['Employment', 'JobSat', 'RemoteWork']].isna().sum())
```

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
## Write your code here
# Convert YearsCodePro to numeric for grouping
years_pro = (
    df_eda['YearsCodePro'] # Select the 'YearsCodePro' column
      .replace({'Less than 1 year': '0.5', 'More than 50 years': '51'}) # Replace special cases with numeric values
      .pipe(pd.to_numeric, errors='coerce') # Convert to numeric, coercing errors to NaN
)

experience_bins = [-float('inf'), 5, 10, 20, float('inf')]
experience_labels = ['0-5', '5-10', '10-20', '>20']
df_eda['ExperienceRange'] = pd.cut( # Create a new column 'ExperienceRange' by binning the 'YearsCodePro' values
    years_pro, bins=experience_bins, labels=experience_labels, # Define the bins and labels for the ranges
    right=True, include_lowest=True # Include the lowest value in the first bin
)
# Calculate the median JobSat for each experience range
median_satisfaction = (
    df_eda.groupby('ExperienceRange', observed=False)['JobSat'] # Group the DataFrame by 'ExperienceRange' and select the 'JobSat' column
          .median()
          .reindex(experience_labels) # Reindex to ensure the order of the experience ranges is maintained
)
print(median_satisfaction)

plt.figure(figsize=(9, 5))
median_satisfaction.plot(kind='bar', color='steelblue')
plt.title('Median Job Satisfaction by Professional Experience')
plt.xlabel('Years of professional coding experience')
plt.ylabel('Median JobSat')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

#### Step 5: Visualize Job Satisfaction


Use a count plot to show the distribution of `JobSat` values. This provides insights into the overall satisfaction levels of respondents.


```python
## Write your code here
plt.figure(figsize=(9,5))
sns.countplot(data=df_eda,x='JobSat',color='seagreen')
plt.title('Distribution of Job Satisfaction')
plt.xlabel('Job Satisfaction score')
plt.ylabel('Number of respondents')
plt.tight_layout()
plt.show()
```

#### Step 6: Analyzing Remote Work Preferences by Job Role


Analyze trends in remote work based on job roles. Use the `RemoteWork` and `Employment` columns to explore preferences and examine if specific job roles prefer remote work more than others.

- Use a count plot to show remote work distribution.

- Cross-tabulate remote work preferences by employment type (e.g., full-time, part-time) and job roles.


```python
## Write your code here
# Choose the job-role column available in this version of the survey
role_column = 'DevType' if 'DevType' in df_eda.columns else 'Employment' # Use 'DevType' if available, otherwise fall back to 'Employment'

plt.figure(figsize=(10,5))
sns.countplot(data=df_eda,x='RemoteWork',color='cornflowerblue')
plt.title('Remote Work Distribution')
plt.xlabel('Remote Work arrangement')
plt.ylabel('Number of respondents')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()

# Cross-tabulate remote work by employment type
remote_by_employment = pd.crosstab( # Create a cross-tabulation of remote work by employment type
    df_eda['Employment'], df_eda['RemoteWork'], normalize='index'
)
print(remote_by_employment.round(3))

# For multi-valued DevType, show the most frequent role categories
role_counts = df_eda[role_column].value_counts().head(10).index # Get the top 10 most frequent roles
role_subset = df_eda[df_eda[role_column].isin(role_counts)] # Create a subset of the DataFrame containing only the top roles
role_remote = pd.crosstab(role_subset[role_column], role_subset['RemoteWork']) # Create a cross-tabulation of remote work by role for the subset
print(role_remote)
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
## Write your code here
# Split the multi-valued language field and count languages by country
language_data = df_eda[['Country', 'LanguageHaveWorkedWith']].dropna().copy() # Create a new DataFrame with only the 'Country' and 'LanguageHaveWorkedWith' columns, dropping rows with NaN values
language_data['Language'] = language_data['LanguageHaveWorkedWith'].str.split(';') # Split the 'LanguageHaveWorkedWith' column into a list of languages
language_data = language_data.explode('Language') # Explode the 'Language' column to create a row for each language
language_data['Language'] = language_data['Language'].str.strip() # Remove leading and trailing whitespace from the 'Language' column

top_countries = df_eda['Country'].value_counts().head(5).index # Get the top 5 countries by number of respondents
language_data = language_data[language_data['Country'].isin(top_countries)] #   Filter the language data to include only the top countries
language_counts = (
    language_data.groupby(['Country', 'Language']).size()
                .reset_index(name='Responses')
)
top_languages = (
    language_counts.groupby('Language')['Responses'].sum()
                  .nlargest(10).index
)
# Create a pivot table for the top languages by country
language_pivot = ( 
    language_counts[language_counts['Language'].isin(top_languages)] # Filter the language counts to include only the top languages
    .pivot(index='Country', columns='Language', values='Responses') # Create a pivot table with countries as rows, languages as columns, and response counts as values
    .fillna(0)
)
print(language_pivot)

plt.figure(figsize=(14, 6))
sns.heatmap(language_pivot, annot=True, fmt='.0f', cmap='YlGnBu')
plt.title('Top Programming Languages by Country')
plt.xlabel('Programming language')
plt.ylabel('Country')
plt.tight_layout()
plt.show()
```

#### Step 8: Correlation Between Experience and Satisfaction


Examine how years of experience (`YearsCodePro`) correlate with job satisfaction (`JobSatPoints_1`). Use a scatter plot to visualize this relationship.


```python
## Write your code here
# Scatter plot and correlation between experience and satisfaction points
experience_sat = pd.DataFrame({ # Create a DataFrame with the relevant columns
    'YearsCodePro_numeric': years_pro, # Convert YearsCodePro to numeric for correlation analysis
    'JobSatPoints_1': pd.to_numeric(df_eda['JobSatPoints_1'], errors='coerce') # Convert JobSatPoints_1 to numeric, coercing errors to NaN
}).dropna() # Drop rows with NaN values in either column

print('Correlation:', experience_sat['YearsCodePro_numeric'].corr( 
    experience_sat['JobSatPoints_1'])) # Print the correlation coefficient

plot_sample = experience_sat.sample(5000, random_state=42) if len(experience_sat) > 5000 else experience_sat # Sample the data for visualization
plt.figure(figsize=(9, 5))
sns.scatterplot(data=plot_sample, x='YearsCodePro_numeric', y='JobSatPoints_1',
                alpha=0.35, s=25)
plt.title('Professional Experience and Job Satisfaction Points')
plt.xlabel('Years of professional coding experience')
plt.ylabel('JobSatPoints_1')
plt.tight_layout()
plt.show()
```

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
## Write your code here
# Cross-tabulate education level and employment type
education_employment = pd.crosstab(
    df_eda['EdLevel'], df_eda['Employment'], normalize='index'
)
print(education_employment.round(3))

# Plot the most common education categories for readability
top_education = df_eda['EdLevel'].value_counts().head(10).index
plot_table = education_employment.loc[education_employment.index.intersection(top_education)]
plot_table.plot(kind='barh', stacked=True, figsize=(12, 7), colormap='tab20')
plt.title('Employment Type by Educational Background')
plt.xlabel('Share of respondents')
plt.ylabel('Education level')
plt.legend(title='Employment', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
