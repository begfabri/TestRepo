### Step 1: Import Required Libraries


```python
import pandas as pd
```

### Step 2: Load the Dataset Into a Dataframe


#### **Read Data**
<p>
The functions below will download the dataset into your browser:
</p>


```python
file_path ="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/UDKAZw-kz18Yj8P6icf_qw/survey-data-duplicates.csv"
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### Step 3. Finding and Removing Duplicates
##### Task 1: Identify duplicate rows in the dataset.


```python
## Write your code here
duplicate_count=df.duplicated().sum()
print(f'Number og duplicate rows: {duplicate_count}')
print(df[df.duplicated()].head())
```

##### Task 2: Remove the duplicate rows from the dataframe.



```python
## Write your code here
df_clean=df.drop_duplicates().copy()
print(f'Rows before removal: {len(df)}')
print(f'Rows after removal: {len(df_clean)}')
print(f'Rows removed: {len(df) - len(df_clean)}')
print(f'Duplicates remaining: {df_clean.duplicated().sum()}')

```

### Step 4: Finding Missing Values
##### Task 3: Find the missing values for all columns.


```python
## Write your code here
missing_values=df_clean.isna().sum().sort_values(ascending=False)
missing_percentage=( missing_values / len(df_clean) * 100).round(2)
missing_summary=pd.DataFrame({
    'Missing values': missing_values,
    'Missing percentage': missing_percentage
})
print(missing_summary[missing_summary['Missing values']>0])
```

##### Task 4: Find out how many rows are missing in the column RemoteWork.


```python
## Write your code here
missing_remote_work=df_clean['RemoteWork'].isna().sum()
print(f'Rows missing RemoteWork: {missing_remote_work}')
```

### Step 5. Imputing Missing Values
##### Task 5: Find the value counts for the column RemoteWork.


```python
## Write your code here
remote_work_counts=df_clean['RemoteWork'].value_counts(dropna=False)
print(remote_work_counts)
```

##### Task 6: Identify the most frequent (majority) value in the RemoteWork column.



```python
## Write your code here
remote_work_mode=df_clean['RemoteWork'].mode(dropna=True)
if remote_work_mode.empty:
    raise ValueError('RemoteWork contains no non-missing values')
majority_Remote_Work=remote_work_mode.iloc[0]
print(f'Most frequent RemoteWork value: {majority_Remote_Work}')
```

##### Task 7: Impute (replace) all the empty rows in the column RemoteWork with the majority value.



```python
## Write your code here
missing_before=df_clean['RemoteWork'].isna().sum()
df_clean['RemoteWork']=df_clean['RemoteWork'].fillna(majority_Remote_Work)
missing_after=df_clean['RemoteWork'].isna().sum()
print('Missing RemoteWork before imputation:',missing_before)
print('Missing RemoteWork after imputation:',missing_after)
```

##### Task 8: Check for any compensation-related columns and describe their distribution.



```python
## Write your code here
#Identify compensation-related columns by name
compensation_columns=[ column for column in df_clean.columns if any(term in column.lower() for term in ('comp','salary','wage'))]
print(f'Compensation related columns; {compensation_columns}')

for column in compensation_columns:
    numeric_values=pd.to_numeric(df_clean[column],errors='coerce') #Convert the column to numeric, coercing errors to NaN
    if numeric_values.notna().any(): #Check if there are any numeric values in the column
        print(f'\nDistribution of {column}:')
        print(numeric_values.describe())
        print(f'Missing values: {numeric_values.isna().sum()}')
    else:
        print(f'\n{column} does not contain numeric values.')
```
