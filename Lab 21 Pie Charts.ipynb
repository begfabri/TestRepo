### Task 1: Visualizing Data Composition with Pie Charts


##### 1.1 Create a Pie Chart of the Top 5 Databases Respondents Want to Work With


In the survey data, the `DatabaseWantToWorkWith` column lists the databases that respondents wish to work with. Let’s visualize the top 5 most-desired databases in a pie chart.



```python
##Write your code here
def top_categories(series, top_n=5, split_multiselect=False):
    values = series.dropna().astype(str).str.strip()
    if split_multiselect:
        values = values.str.split(';').explode().str.strip()
    values = values[values.ne('') & values.ne('nan')]
    counts = values.value_counts()
    top = counts.head(top_n).copy()
    other = counts.iloc[top_n:].sum()
    if other > 0:
        top.loc['Other'] = other
    return top

def pie_from_counts(counts, title, legend_title='Category'):
    counts = counts[counts > 0]
    fig, ax = plt.subplots(figsize=(8, 6))
    wedges, _, _ = ax.pie(counts.values, startangle=90, autopct='%1.1f%%', pctdistance=0.72, wedgeprops={'edgecolor': 'white', 'linewidth': 1})
    ax.legend(wedges, [f'{label} ({value:,})' for label, value in counts.items()], title=legend_title, loc='center left', bbox_to_anchor=(1, 0.5), fontsize=9)
    ax.set_title(title)
    ax.axis('equal')
    plt.tight_layout()
    plt.show()
    return counts

counts=top_categories(survey_df['DatabaseWantToWorkWith'],top_n=5,split_multiselect=True)
pie_from_counts(counts,'Top 5 Preferred Databases', 'Database')

```

The `DevType` column lists the developer types for respondents. We’ll examine the distribution by showing the top 5 developer roles in a pie chart.



```python
##Write your code here
counts=top_categories(survey_df['DevType'],top_n=5,split_multiselect=True)
pie_from_counts(counts,'Top 5 Developer Roles', 'Developer role')
```

##### 1.3 Create a pie chart for the operating systems used by respondents for professional use


The `OpSysProfessional` use column shows the operating systems developers use professionally. Let’s visualize the distribution of the top operating systems in a pie chart.
