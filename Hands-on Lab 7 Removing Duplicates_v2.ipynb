### Step 1: Import Required Libraries


```python
import pandas as pd
```

### Step 2: Load the Dataset into a DataFrame



load the dataset using pd.read_csv()


```python
# Define the URL of the dataset
file_path ="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/UDKAZw-kz18Yj8P6icf_qw/survey-data-duplicates.csv"

# Load the dataset into a DataFrame
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### Step 3: Identifying Duplicate Rows


**Task 1: Identify Duplicate Rows**
  1. Count the number of duplicate rows in the dataset.
  2. Display the first few duplicate rows to understand their structure.


```python
## Write your code here
duplicate_count = df.duplicated().sum()
print(f'Number of duplicate rows: {duplicate_count}')
duplicate_rows=df[df.duplicated()]
print(duplicate_rows.head())
```

### Step 4: Removing Duplicate Rows


**Task 2: Remove Duplicates**
   1. Remove duplicate rows from the dataset using the drop_duplicates() function.
2. Verify the removal by counting the number of duplicate rows after removal .


```python
## Write your code here
# Remove complete duplicate rows and keep the result in a new dataframe
df_clean = df.drop_duplicates().copy()

# Verify that no complete duplicates remain
duplicates_after = df_clean.duplicated().sum()

print("Number of rows before removing duplicates:", len(df))
print("Number of rows after removing duplicates:", len(df_clean))
print("Number of Rows dropped:", len(df) - len(df_clean))
print("Duplicate rows after removal:",duplicates_after)
```

### Step 5: Handling Missing Values


**Task 3: Identify and Handle Missing Values**
   1. Identify missing values for all columns in the dataset.
   2. Choose a column with significant missing values (e.g., EdLevel) and impute with the most frequent value.


```python
## Write your code here
# Count missing values in every column
missing_values=df_clean.isna().sum().sort_values(ascending=False)
print(missing_values[missing_values>0])

# Impute missing EdLevel values with most frequent category
edlevel_mode=df_clean['EdLevel'].mode(dropna=True)
if not edlevel_mode.empty:
    edlevel_fill_value=edlevel_mode.iloc[0]
    missing_edlevel_before=df_clean['EdLevel'].isna().sum()
    df_clean['EdLevel']=df_clean['EdLevel'].fillna(edlevel_fill_value)
    missing_edlevel_after=df_clean['EdLevel'].isna().sum()
    print('EdLevel imputation value:',edlevel_fill_value)
    print('Missing EdLevel before:',missing_edlevel_before)
    print('Missing EdLevel after:',missing_edlevel_after)
else:
    print('EdLevel has no non-missing value available for imputation')
```

### Step 6: Normalizing Compensation Data


**Task 4: Normalize Compensation Data Using ConvertedCompYearly**
   1. Use the ConvertedCompYearly column for compensation analysis as the normalized annual compensation is already provided.
   2. Check for missing values in ConvertedCompYearly and handle them if necessary.


```python
## Write your code here
# ConvertedCompYearly is already an annual, normalized compensation measure.
# Convert it to numeric so it can be used safely in analysis.
df_clean['ConvertedCompYearly'] = pd.to_numeric(
    df_clean['ConvertedCompYearly'], errors='coerce' # coerce will convert non-numeric values to NaN
)

missing_compensation=df_clean['ConvertedCompYearly'].isna().sum() # count the number of missing values in the ConvertedCompYearly column
print('Missing ConvertedCompYearly values:',missing_compensation)
print('Non Missing compensation values:',df_clean['ConvertedCompYearly'].notna().sum())

# Keep missing compensation values as NaN because imputing salary with a mean or median could create an artificial salary for a non-respondent
compensation_summary=df_clean['ConvertedCompYearly'].describe()
print(compensation_summary)
```

### Step 7: Summary and Next Steps


**In this lab, you focused on identifying and removing duplicate rows.**

- You handled missing values by imputing the most frequent value in a chosen column.

- You used ConvertedCompYearly for compensation normalization and handled missing values.

- For further analysis, consider exploring other columns or visualizing the cleaned dataset.


```python
## Write your code here
print('Final dataset shape:',df_clean.shape)
print('Complete duplicates remaining:',df_clean.duplicated().sum())
print('Missing EdLevel values remaining:',df_clean['EdLevel'].isna().sum())
print('Missing ConvertedCompYearly values:',df_clean['ConvertedCompYearly'].isna().sum())

print('The cleaned dataset is ready for exploratory analysis')
```
