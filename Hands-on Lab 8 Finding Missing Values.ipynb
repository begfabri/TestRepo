The functions below will download the dataset into your browser:

### 2. Explore the Dataset
##### Task 1: Display basic information and summary statistics of the dataset.


```python
## Write your code here
df.info()
df.describe()
```

### 3. Finding Missing Values
##### Task 2: Identify missing values for all columns.


```python
## Write your code here
missing_values=df.isna().sum().sort_values(ascending=False)
missing_percentage=(missing_values/ len(df)*100).round(2)
missing_summary=pd.DataFrame({
    'Missing values': missing_values,
    'Missing percentage:': missing_percentage
})
print(missing_summary[missing_summary['Missing values']>0])
```

##### Task 3: Visualize missing values using a heatmap (Using seaborn library).



```python
## Write your code here
# Visualize missing values with a heatmap.
# A sample of rows keeps the plot readable for this large dataset.
plt.figure(figsize=(18, 8))
sample_for_plot = df.isna().sample(1000, random_state=42) if len(df) > 1000 else df.isna() # Sample 1000 rows if the dataset is larger than 1000 rows, otherwise use the entire dataset
sns.heatmap(sample_for_plot, cbar=False, yticklabels=False) # Create a heatmap of missing values, without a color bar and without y-axis tick labels for clarity
plt.title('Missing Values Heatmap (sample of rows)')
plt.xlabel('Columns')
plt.ylabel('Rows')
plt.tight_layout()
plt.show()
```

##### Task 4: Count the number of missing rows for a specific column (e.g., `Employment`).


```python
## Write your code here
missing_employment=df['Employment'].isna().sum()
print('Missing Employment values:',missing_employment)
```

### 4. Imputing Missing Values
##### Task 5: Identify the most frequent (majority) value in a specific column (e.g., `Employment`).


```python
## Write your code here
employment_mode=df['Employment'].mode(dropna=True)
if employment_mode.empty:
    raise ValueError('Employment contains no non-missing values')
most_frequent_employment=employment_mode.iloc[0]
print('Most frequent Employment value:',most_frequent_employment)
```

##### Task 6: Impute missing values in the `Employment` column with the most frequent value.



```python
## Write your code here
missing_before=df['Employment'].isna().sum()
df['Employment']=df['Employment'].fillna(most_frequent_employment)
missing_after=df['Employment'].isna().sum()
print('Missing Employment value before imputation:',missing_before)
print('Missing Employment value after imputation:',missing_after)
```

### 5. Visualizing Imputed Data
##### Task 7: Visualize the distribution of a column after imputation (e.g., `Employment`).


```python
## Write your code here
plt.figure(figsize=(12,6))
order=df['Employment'].value_counts().index
sns.countplot(data=df, y='Employment', order=order, color='steelblue')
plt.title('Employment Distribution After Imputation')
plt.xlabel('Number of respondents')
plt.ylabel('Employment') 
plt.tight_layout()
plt.show()
```
